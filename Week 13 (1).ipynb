```python
import pandas as pd  
import matplotlib.pyplot as mp
import numpy as np
import seaborn as sb
import plotly.express as pe
app = pd.read_excel("applications.xlsx")
```

```python
print(app.head())
```

```python
#### Question 1
app_per = app.groupby('College')['Applicant'].count()
mp.bar(app_per.index, app_per.values, color = 'red') #### Creates the barplot
mp.title('Applicants By College')  ###creates the titles
mp.xlabel('College') ###Creates the x label
mp.ylabel('Applicant Sum') ### creates the y label
mp.show() ### Displays the bar plot
```

```python
#### Seaborn Bar Plot 
sb.countplot(data = app, x = 'College',color = 'blue')
mp.title('Applicants By College')
mp.xlabel('College')
mp.ylabel('Applicant Count')
mp.show()
```

```python
#### Using Plotly 
app_count = app.groupby('College', as_index = False)['Applicant'].count()
fig = pe.bar(app_count, x = 'College', y = 'Applicant', color_discrete_sequence = ['green'], title = 'Applicants by College')
fig.update_layout( xaxis_title = 'College', yaxis_title = 'Applicant Count')
```

```python
#### Question 2
mp.hist(app['SAT_ACT'],
        bins = 10 , color = 'red') 
mp.title('SAT Distribution')
mp.xlabel('Score')
mp.ylabel('Number in range')
mp.show()
```

```python
####The most common scor range is between 1,100 and 1.200
```

```python
###Seaborn 
sb.histplot(data = app, x = 'SAT_ACT',bins = 10 , color = 'blue')
mp.title('SAT Distribution')
mp.xlabel('Score')
mp.ylabel('Number in range')
mp.show()
```

```python
#### Histogram using Plotly 
hist = pe.histogram(app, x = 'SAT_ACT', nbins = 10, color_discrete_sequence = ['green'], title = 'SAT Distribution')
hist.update_layout(xaxis_title = 'SAT Score' , yaxis_title = 'numbe in range')
                   
```

```python
#### Question 3 
mp.scatter(app['Extracurricular_Hours'],app['SAT_ACT'],color = 'red') ### Creates the scatter
mp.title('SAT & Extracurricular Hours')
mp.xlabel('Extra Hours')
mp.ylabel('SAT SCORES')
mp.show()
```

```python
### ANSWER question: No clear relationship is seen as score vary across different hours
```

```python
### Scatter in Seaborn
sb.regplot(data = app, y = 'SAT_ACT' , x = 'Extracurricular_Hours', scatter_kws = {'color':'blue'},ci = None)
mp.title('SAT & Extracurricular Hours')
mp.ylabel('SAT Scores')
mp.xlabel('Extra Hours')
mp.show()
```

```python
### Scatter using Pl
fig = pe.scatter(app, y = 'SAT_ACT', x = 'Extracurricular_Hours', trendline = 'ols', 
color_discrete_sequence = ['green'], title = 'SAT & Extracurricular Hours')
fig.update_layout(yaxis_title = 'SAT Score', xaxis_title  = 'Extra Hours')
```
