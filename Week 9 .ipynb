```python
####Select Sales Data
import sqlite3
import pandas as pd
conn = sqlite3.connect("My_Data.db")
cursor = conn.cursor()
```

```python
pd.read_csv('sales.csv').to_sql('sales',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select sale_date , product, price From sales 
''',conn)
```

```python
###Where
pd.read_csv('hr_employees.csv').to_sql('employees',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select name, department, salary, years_experience, city
from employees
where city = 'Chicago' AND years_experience > 10
''',conn)
```

```python
pd.read_sql_query('''
select listing_id, neighborhood, bedrooms, sqft, list_price
from listings
order by sqft desc limit 3
''', conn)
```

```python
###Grouping
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select  sum(impressions) As Total_Impressions 
from campaigns
group by week
''',conn)
```

```python
###Having
pd.read_csv('restaurant_orders.csv').to_sql('orders',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select order_date, sum(order_total) as Date_total
From orders 
group by order_date
Having sum(order_total) > 400
''',conn)
```

```python
###aggregate
pd.read_csv('clinic_visits.csv').to_sql('visits',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select doctor, count(*) as total_visits, 
        round(AVG(patient_age),0) as Average_Age,
        Max(wait_minutes) AS Highest_wait
    from visits
    group by doctor
    ''',conn)
```

```python
###Join
pd.read_csv('bank_customers.csv').to_sql('customers',conn,if_exists = 'replace',index = False)
pd.read_csv('bank_transactions.csv').to_sql('transactions',conn,if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
select branch, t.transaction_type, sum(t.amount) as Total_Amount
from transactions as t
join customers as c on t.customer_id = c.customer_id
group by branch 
''',conn)

```
